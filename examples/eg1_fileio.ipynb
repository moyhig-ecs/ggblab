```python
from ggblab import GeoGebra
import xml.etree.ElementTree as ET
import xmlschema
```

```python
# initialize base class, not open GeoGebra Widget
ggb = GeoGebra()
```

```python
# load from .ggb (zipped, and base64 encoded)
c = ggb.construction.load('2025_06_08.ggb')
```

```python
# parse loaded xml as python dict
o = c.ggb_schema.decode(c.geogebra_xml)
```

```python
# geogebra user data structure
for e in o:
    print(e)
```

```python
# basicaly geogebra save as a file with title
o['@title'], ggb.construction.source_file
```

```python
# on default, not overwrite but assgin revision number at the tail of the base of filename
c.save()
```

```python
# to overwrite
c.save(overwrite=True)
```
