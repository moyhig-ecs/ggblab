# eg1 (v2) — reading a `.ggb` file

No server and no applet are needed. Version 1 read the file with `ggb.file.load(...)`; version 2 has two functions: `read_ggb` returns the XML text, and `ConstructionIO.from_ggb_file` returns a data frame.

```python
from ggblab_extra import ConstructionIO, read_ggb
from ggblab.schema import get_schema
from ggblab.xml_errata import construction_xml
```

```python
# load from .ggb (zipped, and base64 encoded)
xml = read_ggb('2025_06_08.ggb')
print(len(xml), 'characters')
```

```python
# parse the construction as a python dict
o = get_schema().decode(construction_xml(xml))
```

```python
# geogebra user data structure
for e in o:
    print(e)
```

```python
# the title, and the number of elements and commands
o['@title'], len(o['element']), len(o['command'])
```

```python
# the same file as a data frame: one row per object
df = ConstructionIO.from_ggb_file('2025_06_08.ggb')
print(df.shape)
df.head(8)
```

## Saving

Version 1 wrote the construction back to a `.ggb` file (`c.save()`). Version 2 does not write `.ggb` files. The data frame can be saved as Parquet or JSON.

```python
import tempfile
out = ConstructionIO.save_dataframe(df, fmt='parquet', out_dir=tempfile.mkdtemp())
print(out)
print(ConstructionIO.from_parquet(out).shape)
print('DONE')
```
