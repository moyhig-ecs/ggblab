# eg2 (v2) — the construction XML as a Python dict, offline

No server and no applet are needed. The GeoGebra XML schema is shipped with the package. `decode` turns XML into a dict whose attribute keys start with `@`; `encode` turns a dict back into XML.

```python
import xmlschema
from ggblab_extra import read_ggb
from ggblab.schema import get_schema
from ggblab.xml_errata import construction_xml
```

```python
# load from .ggb (zipped, and base64 encoded)
xml = read_ggb('2025_06_08.ggb')
sch = get_schema()
```

```python
# parse the construction as a python dict
o = sch.decode(construction_xml(xml))
```

```python
# geogebra user data structure
for e in o:
    print(e)
```

```python
# list of all object names
labels = [e['@label'] for e in o['element']]
labels
```

```python
labels.index('A')
```

```python
o['element'][1]
```

```python
# construct xml attributes from dict
x = xmlschema.etree_tostring(sch.schema.encode(o['element'][1], 'element'))
print(x)
```

```python
# encode again to dict from xml
o2 = sch.decode(x)
print(o2 == o['element'][1])
print('DONE')
o2
```
