```python
import sys, re, time; sys.path.insert(0, '/Users/manabu/work/ggblab-replay')
import ggblab.host.html_host as H; H.DEPLOY = 'https://cdn.geogebra.org/apps/deployggb.js'
from ggblab import GeoGebra
from ggblab.host.base import Verb
from ggblab.parse import parse_cell
from ggblab.adapter import plan, apply
print('verbs', [v.value for v in Verb])
g = GeoGebra(appName='suite'); g
```

```python
CELL = '''@ggb :const :new
@ggb O=(0,0)
@ggb Circle(:O, 1)
@ggb P=(1, 0)
@ggb Q=(-1, 0)
@ggb Segment(:P, :Q)
@ggb :const :new
@ggb l_CA = Line(:P, :Q)
@ggb M = Midpoint(P, Q)
'''
c = parse_cell(CELL); print('plan', [type(s).__name__ for s in plan(c)]); print('deps', c.dependencies())
t0 = time.time(); g.command('Z = (9, 9)', timeout=60)              # pre-existing object: must vanish at :const :new (LABELS_AFTER must not contain Z)
out = apply(g, c, timeout=60); xml = g.xml(timeout=60)
labels = re.findall(r'<element type="\w+" label="([^"]+)"', xml)
print('apply', round(time.time()-t0, 2), 's; replies', [str(o)[:60] for o in out]); print('LABELS_AFTER', labels)
open('/private/tmp/claude-501/-Users-manabu-lancedb-rag/fd32421b-f579-4fc8-aad6-f0fe35d00c39/scratchpad/gate2/new_verb_xml.txt', 'w').write(xml)
```
