```python
await ggb.command("O = (0, 0)")
await ggb.command("c1 = Circle(O, 1)")
await ggb.command("A = (1, 0)")
await ggb.command("c2 = Circle(A, 1)")
await ggb.command("l1 = {Intersect(c1, c2)}")
await ggb.command("a = Length(l1)")
```

## with ipywidgets

```python
import ipywidgets as widgets
```

```python
import ipywidgets as widgets
from IPython.display import display

# IntSliderの作成
int_slider = widgets.IntSlider(
    value=5,          # 初期値
    min=0,            # 最小値
    max=10,           # 最大値
    step=1,           # 刻み幅
    description='整数:', # ラベル
    orientation='horizontal' # 'horizontal' (水平) か 'vertical' (垂直)
)

# 表示
display(int_slider)

# 値の取得方法 (別のセルなどで実行)
# print(int_slider.value)
```

```python
label = widgets.Label(value='a') #ggb.comm.shared_objects['a'])
```

```python
display(label)
```

```python
async def on_shared_update(changes):
    # await asyncio.sleep(0)
    label.value = changes['a']
```

```python
ggb.comm.add_shared_listener(on_shared_update)
```

```python
ggb.comm.logs
```

## clear

```python
ggb.comm.remove_shared_listener(on_shared_update)
```

```python
ggb.comm.clear_shared_listner()
```

```python
r = await ggb.listen('a', False)
```
