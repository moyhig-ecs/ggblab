```python
# open GeoGebra Widget on left-side
# ggb = await GeoGebra().init()
ggb = await GeoGebra().init(use_vscode=False)
```

```python
# case 1 of interactions: algebraic command
# [GeoGebra Manual :: GeoGebra Manual]
# (https://geogebra.github.io/docs/manual/)
```

```python
await ggb.command("O = (0, 0)")
await ggb.command("c1 = Circle(O, 1)")
await ggb.command("A = (1, 0)")
await ggb.command("c2 = Circle(A, 1)")
await ggb.command("l1 = {Intersect(c1, c2)}")
await ggb.command("a = Length(l1)")
```

```python
ggb.comm.logs
```

```python
# moveing Point 'A', 'a' corresponds number of intersecting points
# but GeoGebra returns 1 even when no intersection...
r = await ggb.listen('a')

await asyncio.sleep(.01)
ggb.comm.shared_objects
```

## with ipywidgets

```python
import ipywidgets as widgets
```

```python
label = widgets.Label(value=ggb.comm.shared_objects['a'])
```

```python
display(label)
```

```python
async def on_shared_update(changes):
    # await asyncio.sleep(0)
    label.value = changes['a']
```

```python
ggb.comm.add_shared_listener(on_shared_update)
```

```python
ggb.comm.logs
```

## clear

```python
ggb.comm.remove_shared_listener(on_shared_update)
```

```python
ggb.comm.clear_shared_listner()
```

```python
r = await ggb.listen('a', False)
```
