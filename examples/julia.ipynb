```julia
using Sockets, JSON

function send_to_bridge(payload::Dict; host="127.0.0.1", port=8765)
    sock = connect(host, port)
    try
        write(sock, JSON.json(payload) * "\n")
        flush(sock)
        resp = readline(sock)
        return JSON.parse(resp)
    finally
        close(sock)
    end
end

# 使い方
send_to_bridge(Dict("op"=>"ping"))
```

```julia
using IJulia, JSON

const TARGET_NAME = "jupyter.ggblab"

function on_open(comm, msg)
    println("ggblab comm opened (manual)")
    comm.on_msg(function(m)
        data = m["content"]["data"]
        payload = isa(data, String) ? JSON.parse(data) : data
        id = get(payload, "id", nothing)
        if get(payload, "type", "") == "function"
            value = "simulated-result-for-" * string(payload["payload"]["name"])
            reply = Dict("type" => "value", "id" => id, "payload" => Dict("value" => value))
        else
            reply = Dict("type" => "error", "id" => id, "payload" => Dict("message" => "unsupported type"))
        end
        # Use IJulia.send / comm.send fallback to reply
        try
            IJulia.send(comm, JSON.json(reply))
        catch
            try
                comm.send(JSON.json(reply))
            catch ee
                @warn("Failed to send reply in manual on_open: " * string(ee))
            end
        end
    end)
end

if hasproperty(IJulia, :register_comm)
    IJulia.register_comm(TARGET_NAME, on_open)
elseif hasproperty(IJulia, :install_comm)
    IJulia.install_comm(TARGET_NAME, on_open)
else
    println("No registration API available in this IJulia build; the frontend cannot open comms to Julia automatically.")
end
```

```julia
include("open_ggblab_control.jl")   # defines open_ggblab_control()
using JSON
using UUIDs

# Open control comm (returns Comm or nothing)
c = open_ggblab_control()
if c === nothing
    error("Couldn't open control comm")
end

# Simple reply waiter registry
const _REPLIES = Dict{String,Channel{Any}}()

function _on_msg(comm, m)
    data = m["content"]["data"]
    payload = isa(data, String) ? JSON.parse(data) : data
    id = get(payload, "id", nothing)
    if id !== nothing && haskey(_REPLIES, id)
        put!(_REPLIES[id], payload)
    else
        println("unsolicited:", payload)
    end
end

c.on_msg(m -> _on_msg(c, m))

# Call a function synchronously and wait for reply
function call_function_sync(comm, name, args; timeout=5.0)
    id = string(UUIDs.uuid4())
    req = Dict("type"=>"function", "id"=>id, "payload"=>Dict("name"=>name, "args"=>args))
    ch = Channel{Any}(1)
    _REPLIES[id] = ch
    try
        payload = JSON.json(req)
        try
            IJulia.send_comm(comm, payload)
        catch
            comm.send(payload)
        end
        # wait with timeout
        t_start = time()
        while true
            if isready(ch)
                resp = take!(ch)
                delete!(_REPLIES, id)
                return resp
            elseif time() - t_start > timeout
                delete!(_REPLIES, id)
                error("timeout waiting for reply")
            else
                sleep(0.01)
            end
        end
    finally
        if haskey(_REPLIES, id)
            delete!(_REPLIES, id)
        end
    end
end

# Example usage:
resp = call_function_sync(c, "getValue", ["A"])
println("reply: ", resp)
```
