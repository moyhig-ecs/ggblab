```python
c = ggb.file.load('2025_02_03.ggb')
```

```python
o = c.ggb_schema.decode(c.geogebra_xml)
```

```python
r = await ggb.function("getAllObjectNames")
for _n in r:# set(r) - {'xOyPlane', 'l7', 'l8', 'l9', 'l3', 'l2', 'l1', 'l0', 'l6', 'l4', 'l5'}:
    # print(_n)
    try:
        _i = [e['@label'] for e in o['element']].index(_n)
        _e = o['element'][_i]
        for _k in ['animation', 'pointSize', 'pointStyle', 'coords', 'eigenvectors', 'eqnStyle', 'matrix', 'show']:
            _e.pop(_k, None)
        _x = xmlschema.etree_tostring(c.ggb_schema.encode(_e, 'element'))
        # print(_x)
        await ggb.function("evalXML", [_x])
    except ValueError as e:
        print(e)
        pass
```
