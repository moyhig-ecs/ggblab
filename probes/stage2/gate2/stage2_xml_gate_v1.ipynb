```python
import json, time, asyncio
import ggblab; print('v1', ggblab.__version__, ggblab.__file__)
from ggblab import GeoGebra
GROUPS = json.load(open('/private/tmp/claude-501/-Users-manabu-lancedb-rag/fd32421b-f579-4fc8-aad6-f0fe35d00c39/scratchpad/gate2/groups.json'))
ggb = await GeoGebra().init()   # opens the labextension panel (current route: comm); the applet + comm come up ~17 s later
t0 = time.time(); ver = None
for i in range(90):
    try:
        ver = await ggb.function('getVersion'); break
    except Exception as e:
        await asyncio.sleep(1)
print('comm active after', round(time.time()-t0, 1), 's; applet version', ver)
```
