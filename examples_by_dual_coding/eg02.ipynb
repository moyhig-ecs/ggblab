```python
from sympy import Matrix, cos, sin, symbols, pi

# シンボルの定義
theta, dx, dy, sx, sy = symbols('theta dx dy sx sy')
x, y = symbols('x y')

# 1. 変換行列の定義 (3x3)
# 回転行列 (Rotation)
R = Matrix([
    [cos(theta), -sin(theta), 0],
    [sin(theta),  cos(theta), 0],
    [0,           0,          1]
])

# 平行移動行列 (Translation)
T = Matrix([
    [1, 0, dx],
    [0, 1, dy],
    [0, 0, 1]
])

# スケーリング行列 (Scaling)
S = Matrix([
    [sx, 0, 0],
    [0, sy, 0],
    [0, 0, 1]
])

# 2. 点の定義 (同次座標系: [x, y, 1])
point = Matrix([x, y, 1])

# 3. 変換の適用 (行列の積)
# 例：回転 -> 平行移動
transformed_point = T * R * point

print("変換後の点:")
print(transformed_point)

# 具体的な値を代入する場合
final_point = transformed_point.subs({theta: pi/4, dx: 1, dy: 2, x: 10, y: 0})
print("\n具体的な値(θ=45°, dx=1, dy=2, 10,0):")
print(final_point.evalf())
```
