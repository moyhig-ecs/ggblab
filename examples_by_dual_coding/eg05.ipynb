```python
from sympy.parsing.sympy_parser import parse_expr, standard_transformations, function_exponentiation, implicit_multiplication_application, convert_xor
from sympy import pretty, init_printing

# 1. パースの設定 ( ^ -> ** への変換)
transformations = standard_transformations + (function_exponentiation, implicit_multiplication_application, convert_xor, )

# サンプル式: 2x^2 + 3y*z
expr_str = "a x^2 + b x y + c y^2 + d x + e y + f" 

# パースの実行 (implicit_multiplicationを有効にすると、2xが2*xと認識される)
# 注: transformations="all" を使うとさらに自動化が進む
expr = parse_expr(expr_str, transformations=transformations)
```
